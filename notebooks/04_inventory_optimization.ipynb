```python
import pandas as pd
import numpy as np
forecast_df = pd.read_csv(
    "../outputs/forecasts/office_supplies_12_month_forecast.csv"
)

forecast_df['Month'] = pd.to_datetime(forecast_df['Month'])
forecast_df.head()
```

```python
avg_monthly_demand = forecast_df['Forecasted_Quantity'].mean()
avg_monthly_demand
```

```python
historical_df = pd.read_csv("../data/processed/monthly_category_demand.csv")
office_hist = historical_df[
    historical_df['Category'] == 'Office Supplies'
]['Quantity']

demand_std = office_hist.std()
demand_std
```

```python
lead_time_months = 1
service_level = 0.95      # 95% service level
z_score = 1.65            # standard for 95%
```

```python
safety_stock = z_score * demand_std * np.sqrt(lead_time_months)
safety_stock
```

```python
reorder_point = (avg_monthly_demand * lead_time_months) + safety_stock
reorder_point
```

```python
inventory_policy = pd.DataFrame({
    "Metric": [
        "Average Monthly Demand",
        "Demand Std Dev",
        "Lead Time (months)",
        "Safety Stock",
        "Reorder Point"
    ],
    "Value": [
        round(avg_monthly_demand, 0),
        round(demand_std, 0),
        lead_time_months,
        round(safety_stock, 0),
        round(reorder_point, 0)
    ]
})

inventory_policy
```

```python
inventory_policy.to_csv(
    "../outputs/inventory_policy_office_supplies.csv",
    index=False
)
```
