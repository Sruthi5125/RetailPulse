```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
# Load dataset
file_path = "../data/raw/global_superstore_2016.xlsx"
df = pd.read_excel(file_path)

df.head()
```

```python
df.shape
df.info()
df.describe()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df[['Order Date', 'Ship Date']].dtypes
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python
monthly_demand = (
    df.groupby(['Month', 'Category'])['Quantity']
      .sum()
      .reset_index()
)

monthly_demand.head()
```

```python
monthly_demand['Month'] = monthly_demand['Month'].dt.to_timestamp()
```

```python
monthly_demand.isnull().sum()
```

```python
monthly_demand.describe()
```

```python
monthly_demand.to_csv(
    "../data/processed/monthly_category_demand.csv",
    index=False
)
```
